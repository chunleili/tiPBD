```python
# lame parameter transfer from mu,nu to E
# input: mu,nu
# output: E
def mu_nu_to_E(mu, nu):
    """将拉梅参数μ和ν转换为杨氏模量E
    Args:
        mu: 剪切模量
        nu: 泊松比
    Returns:
        E: 杨氏模量
    """
    return 2 * mu * (1 + nu)

mu = 1e6
nu = 0.499
E = mu_nu_to_E(mu, nu)
print(f"{E:.5e}")
```
